```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
threads = [1,2,4,8,16,20]
speedup_v1 = [1.00,1.91,3.22,3.92,5.62,6.87]
speedup_v2 = [1.00,1.92,2.56,3.21,6.80,7.56]
```

```python
fig, ax = plt.subplots()

ax.plot(threads, speedup_v1, marker='o', label='Version 1', color='blue')
ax.plot(threads, speedup_v2, marker='o', label='Version 2', color='red')
ax.plot(threads,threads,"--",label = "Ideal", color = "green")
```

```python

ax.set_xlabel("Number of threads")
ax.set_ylabel("Speedup")
ax.set_title("Mandelbrot Speedup")
ax.set_xticks(threads)
ax.grid(True)
ax.legend()
```

```python
fig.savefig("speedup.png", dpi=400)
fig.show()
```
